# LightGBM Evaluation Report

Loads the LightGBM model + scaler, evaluates on `splits_2/test.csv`, and outputs:
- `classification_report.csv`
- `confusion_matrix.png`
- `f1_scores.png`


```python
import json
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

import lightgbm as lgb
```

```python
def load_label_mapping(path: Path):
    with path.open("r", encoding="utf-8") as fh:
        mapping = json.load(fh)
    return [mapping[str(i)] for i in range(len(mapping))]


def load_features(csv_path: Path):
    df = pd.read_csv(csv_path)
    if "label" not in df.columns:
        raise ValueError("CSV must include a 'label' column.")
    feature_cols = [c for c in df.columns if c not in ("label", "timestamp", "run_id")]
    return df, feature_cols


def plot_confusion(cm, labels, out_path: Path, title: str):
    plt.figure(figsize=(12, 10))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=labels,
        yticklabels=labels,
        cbar=False,
    )
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.tight_layout()
    plt.savefig(out_path, dpi=150)
    plt.close()
```

```python
base_dir = Path(__file__).resolve().parents[2]
detector_dir = base_dir / "detector_system_ml"
datasets_dir = base_dir / "ml_system" / "datasets"

model_path = detector_dir / "lightgbm_model_2.txt"
scaler_path = detector_dir / "feature_scaler.pkl"
label_path = detector_dir / "label_mapping.json"
test_csv = datasets_dir / "splits_2" / "test.csv"

out_dir = Path(__file__).resolve().parent
out_dir.mkdir(parents=True, exist_ok=True)
```

```python
labels = load_label_mapping(label_path)
df, feature_cols = load_features(test_csv)

X = df[feature_cols].values
y_true = df["label"].values

scaler = joblib.load(scaler_path)
X_scaled = scaler.transform(X)
```

```python
model = lgb.Booster(model_file=str(model_path))
y_proba = model.predict(X_scaled)
y_pred_idx = np.argmax(y_proba, axis=1)
y_pred = np.array([labels[i] for i in y_pred_idx])
```

```python
report = classification_report(
    y_true, y_pred, labels=labels, output_dict=True, zero_division=0
)
report_df = pd.DataFrame(report).T
report_path = out_dir / "classification_report.csv"
report_df.to_csv(report_path, index=True)
report_df
```

```python
cm = confusion_matrix(y_true, y_pred, labels=labels)
cm_path = out_dir / "confusion_matrix.png"
plot_confusion(cm, labels, cm_path, "Confusion Matrix (Test)")
cm
```

```python
f1_scores = report_df.loc[labels, "f1-score"]
plt.figure(figsize=(12, 4))
sns.barplot(x=f1_scores.index, y=f1_scores.values, color="steelblue")
plt.title("Per-class F1-score (Test)")
plt.ylabel("F1-score")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig(out_dir / "f1_scores.png", dpi=150)
plt.close()

print("Saved:")
print(report_path)
print(cm_path)
print(out_dir / "f1_scores.png")
```
